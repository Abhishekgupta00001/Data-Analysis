```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Aman", "Rahul", "Priya", "Abhishek", "Neha"],
    "Python": [85, 70, 92, 88, 78],
    "SQL": [80, 75, 89, 90, 82],
    "Excel": [90, 80, 85, 95, 88]
}

df = pd.DataFrame(data)

df
```

```python
# Average
```

```python
df = pd.DataFrame(data)
df["Total"] = df["Python"] + df["SQL"] + df["Excel"]

df["Average"] = df["Total"] / 3

df
```

```python
top_student = df.loc[df["Total"].idxmax()]

print("Top Student:", top_student["Name"])
print("Total Marks:", top_student["Total"])
print("Average Marks:", round(top_student["Average"], 2))
```

```python
#Graph banao
```

```python
plt.bar(df["Name"], df["Average"])

plt.title("Student Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.xticks(rotation=30)
plt.show()
```

```python
#Class ka average nikalo
```

```python
print("Class Average:", round(df["Average"].mean(), 2))
print("Highest Average:", df["Average"].max())
print("Lowest Average:", df["Average"].min())
```
